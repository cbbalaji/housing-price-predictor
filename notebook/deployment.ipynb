```python
from sagemaker.sklearn.model import SKLearnModel
from sagemaker.serializers import JSONSerializer
from sagemaker.deserializers import JSONDeserializer
```

```python
role = sagemaker.get_execution_role()
model_dt = "s3://sagemaker-us-east-2-133080914359/sklearn-training/output/sklearn-model-training-2025-09-14-21-16-10-302/output/model.tar.gz"
```

```python
session = sagemaker.Session()
```

```python
!aws s3 ls s3://sagemaker-us-east-2-133080914359/sklearn-training/output/sklearn-model-training-2025-09-14-21-16-10-302/output/model.tar.gz
```

```python
model = SKLearnModel(
    model_data=model_dt,
    role=role,
    entry_point='inference.py',
    framework_version='0.23-1',
    source_dir='../scripts',
    sagemaker_session=session
)
```

```python
predictor = model.deploy(
    instance_type='ml.m5.large',
    initial_instance_count=1,
    endpoint_name='house-price-predictor-ver0',
    serializer=JSONSerializer(),
    deserializer=JSONDeserializer()
)
```

```python
test_input = {
    "housing_median_age": 28,
    "median_income": 5.3,
    "population_per_household": 3.1,
    "rooms_per_household": 6.0,
    "bedrooms_per_room": 0.15,
    "loc_cluster": 2
}


response = predictor.predict(test_input)
print("Prediction:", response)
```
